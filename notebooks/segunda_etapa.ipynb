## 2. Definição da variável-alvo (Target)

A variável-alvo do problema é **`nps_score`**, que representa a nota de satisfação atribuída pelo cliente após sua experiência de compra. Essa variável varia de 0 a 10.

Ela foi escolhida porque é o principal indicador disponibilizado pela empresa para mensurar a percepção do cliente sobre toda a jornada, incluindo pedido, entrega e atendimento.

A coleta ocorre **após o encerramento da experiência de compra**, quando o cliente já recebeu o pedido e, caso necessário, interagiu com o atendimento. Portanto, o score representa uma avaliação final da experiência.

Para análises de negócio, as notas podem ser agrupadas em:

- **Detratores:** notas de 0 a 6;
- **Passivos:** notas de 7 e 8;
- **Promotores:** notas de 9 e 10.

É importante não confundir a nota individual (`nps_score`) com o NPS agregado da empresa, que é calculado pela diferença entre o percentual de promotores e o percentual de detratores.

Também existe risco de uso inadequado da variável: para prever a satisfação de forma preventiva, devem ser usadas apenas informações disponíveis antes da pesquisa de NPS. Variáveis conhecidas somente após a jornada, como recompra em 30 dias (`repeat_purchase_30d`), podem causar vazamento de informação e gerar previsões artificialmente boas.
