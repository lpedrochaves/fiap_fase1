# Primeira Etapa
1. Entendimento do negócio:  nessa primeira etapa, queremos exercitar o seu 
pensamento analítico, não código. Nos traga a resposta para as seguintes 
perguntas de negócio: 
- Qual problema de negócio está sendo resolvido? 
    - R: Alguns clientes não estão satisfeitos, mesmo com dados operacionais sem muita discrepância dos satisfeitos. 
- Por que o NPS é importante para um e-commerce? 
    - R: Ele Pode agir como um termômetro de todo fluxo comercial de uma empresa. 
-  Quais áreas poderiam se beneficiar desses insights? Exemplos: 
    - R: Atendimento, Produto, Financeiro etc.

Como o NPS impacta: 
- Recompra; 
    - Clientes satisfeitos, tendem a retornar, mas insatisfeitos dificilmente são recuperados
- Boca a boca; 
    - A satisfação ou insatisfação pode ser crucial para um cliente novo comprar ou não um produto
- Market share em e-commerce.
    - A concorrência pode se aproveitar e garantir uma fatia maior de mercado.

Quais indicadores de mercado poderiam complementar essa análise? 
- NPS
- SLA logístico
- Concorrência.
